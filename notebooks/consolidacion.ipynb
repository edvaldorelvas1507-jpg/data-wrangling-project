# Consolidación de datos: Apple, Amazon y Microsoft

Junta las tablas de emisiones (Scope 1, 2 y 3) e ingresos de las tres empresas en un solo dataset.

**Entradas** (carpeta `data/raw/`, no se suben a GitHub):
- `amazon_microsoft.xlsx`: Amazon y Microsoft, 2020–2024 (ya trae la columna `empresa`).
- `tablas_amazon_apple.xlsx`: hoja `Apple`, 2021–2024 (no trae la columna `empresa`).

**Salida:** `data/clean/emisiones_consolidado.csv`.

Huecos conocidos (no se inventan): Apple 2020, Apple 2022 Scope 2 location-based, Amazon Scope 2 location-based (todos los años). El Scope 1 de Apple es 55.200 en los cuatro años: pendiente de verificar.

```python
import pandas as pd
import os
```

## 1. Leer las tablas

```python
amazon_microsoft = pd.read_excel("../data/raw/amazon_microsoft.xlsx")
apple = pd.read_excel("../data/raw/tablas_amazon_apple.xlsx", sheet_name="Apple")

print(amazon_microsoft.shape, apple.shape)
amazon_microsoft.head()
```

## 2. Añadir la columna `empresa` a Apple
La tabla de Apple no la tiene, y la necesitamos para poder juntar las tablas.

```python
apple.insert(0, "empresa", "Apple")
apple.head()
```

## 3. Juntar las tablas y ordenar

```python
df = pd.concat([apple, amazon_microsoft], ignore_index=True)
df = df.sort_values(["empresa", "anio"]).reset_index(drop=True)
df
```

## 4. Comprobaciones rápidas

```python
# años disponibles por empresa (los huecos se ven aquí)
print(df.groupby("empresa")["anio"].apply(list))

# valores vacíos por columna
df.isna().sum()
```

## 5. Guardar el dataset consolidado

```python
os.makedirs("../data/clean", exist_ok=True)
df.to_csv("../data/clean/emisiones_consolidado.csv", index=False)
```
